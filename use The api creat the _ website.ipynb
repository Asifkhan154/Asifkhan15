```python
from fastapi import FastAPI
from pydantic import BaseModel, Field, EmailStr
from typing import Literal
import nest_asyncio
import uvicorn
from pyngrok import ngrok
import threading
import time
```

```python
from google.colab.output import eval_js

app = FastAPI(
    title="froduct odaring app",
    description="API demo",
    version="1.0"
)

class order (BaseModel):
  customar_name:str=Field (...,min_length = 3, max_length =50)
  email : EmailStr
  quantity : int=Field(...,ge=1)
  price:float = Field (...,gt=0)

@app.post("/place_order")
def place_order(order:order):
    return {
        "message":"order accepted",
        "customar":order.customar_name,
        "email":order.email,
        "quantity":order.quantity,
        "price":order.price,
        "total":order.quantity*order.price
        }

nest_asyncio.apply()

def run():
    uvicorn.run(app,
                host='0.0.0.0',
                port=8081)

thread = threading.Thread(
    target=run
)

thread.start()

time.sleep(3)

public_url =eval_js("google.colab.kernel.proxyPort(8081)")
print("="*60)
print()
print("public_URL:")
print()
print(public_url)
print()
print("Swagger UI")
print(public_url+"/docs")
print()
print("redoc")
print(public_url+"/redoc")
print()
print("openai json")
print(public_url+"/openai.json")
print()
print("="*60)
```
