pivot table ka matlab hota hai data ko acche trah se nichodna

```python
data={
    'region':['east','west','east','west','east','west'],
    'products':['A','B','C','D','E','F'],
    'sales':[100,200,300,400,500,600]
}
```

```python
df=pd.DataFrame(data)

pivot_table = df.pivot_table(values='sales',index='region',columns='products',aggfunc='sum')
display(pivot_table)
```

```python
data={
    'region':['east','west','east','west','east','west','esat','west'],
    'products':['A','B','C','D','E','F','g','h'],
    'sales': [100, 150, 200, 250, 300, 350, 200, 100]
}
df=pd.DataFrame(data)

pivot_table = df.pivot_table(values='sales',index='region',columns='products',aggfunc='sum',fill_value=0)
pivot_table
```

```python
data = {
    'region': ['east', 'west', 'east', 'west', 'east', 'west', 'east', 'east'],
    'product': ['A', 'A', 'B', 'B', 'C', 'C', 'A', 'D'],
    'sales': [100, 150, 200, 250, 300, 350, 200, 100]
}

df = pd.DataFrame(data)

pivot_table = df.pivot_table(values='sales', index='region', columns='product', aggfunc='sum', fill_value=0)
pivot_table
```

```python
data = {
    'region': ['east', 'west'],
    'product1': [100, 150],
    'product2': [200, 250],
    'product3': [300, 350]
}

df = pd.DataFrame(data)

#melt the dataframe
melted_df = pd.melt(df, id_vars=['region'], value_vars=['product1', 'product2', 'product3'], var_name='product', value_name='sales')
print(melted_df)
```
